```python
df["Cuisines"]=df["Cuisines"].fillna(df["Cuisines"].mode()[0])
```

```python
drop_columns=['Restaurant ID',
 'Restaurant Name',
 'Address',
 'Locality',
 'Locality Verbose',
 'Rating color',
 'Rating text',
 'Switch to order menu']


              
```

```python
df=df.drop(columns=drop_columns)
```

```python
df.head()
```

```python
df.shape
```

```python
X=df.drop('Aggregate rating',axis=1)
y=df['Aggregate rating']
print("features : ")
print(X.columns.tolist())
print("\n Target :")
print(y.name)
```

```python
categorical_columns = X.select_dtypes(include="str").columns

print("Categorical Columns:")
print(categorical_columns.tolist())
```

```python
X = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)

print(X.head())
print("New shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_model=LinearRegression()
```

```python
linear_model.fit(X_train,y_train)
print("linear model run successfully !")
```

```python
y_pred=linear_model.predict(X_test)
```

```python
print(y_pred)
```

```python
from sklearn.metrics import mean_squared_error ,r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
```

```python
print(mse)
```

```python
r2 = r2_score(y_test, y_pred)
```

```python
print(r2)
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=10
)

tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
tree_pred = tree_model.predict(X_test)

tree_mse = mean_squared_error(y_test, tree_pred)
tree_r2 = r2_score(y_test, tree_pred)

print("Decision Tree MSE:", tree_mse)
print("Decision Tree R2 Score:", tree_r2)
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": tree_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance.head(10))
```

```python
comparison = pd.DataFrame({
    "Actual Rating": y_test.iloc[:10].values,
    "Predicted Rating": tree_pred[:10]
})

print(comparison)
```

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Decision Tree Regression"],
    "MSE": [mse, tree_mse],
    "R2 Score": [r2, tree_r2]
})

print(results)

if tree_r2 > r2:
    print("\nBest Model: Decision Tree Regression")
else:
    print("\nBest Model: Linear Regression")
    
```

```python
print("========== FINAL RESULT ==========")

if tree_r2 > r2:
    best_model = "Decision Tree Regression"
    best_mse = tree_mse
    best_r2 = tree_r2
else:
    best_model = "Linear Regression"
    best_mse = mse
    best_r2 = r2

print("Best Model:", best_model)
print("MSE:", best_mse)
print("R2 Score:", best_r2)

print("\n========== CONCLUSION ==========")

print(
    f"The {best_model} performed better based on the R2 Score "
    f"and MSE comparison. The model was used to predict restaurant "
    f"Aggregate Rating based on the available restaurant features."
)
```
